```python
##Assignment1: Write a function that takes two integers as input and returns their division. Use try, except, and finally blocks to handle division by zero and print an appropriate message.
def except_division(a,b):
    try:
        return a / b
    except ZeroDivisionError as e:
        print(f'error: {e}')
except_division(9,0)
```

```python
##Assignment2: Write a function that reads the contents of a file named `data.txt`. Use try, except, and finally blocks to handle file not found errors and ensure the file is properly closed.
def read_file(filename):
    try:
        with open(filename,'r') as file:
            content = file.read()
            print(content)
    except FileNotFoundError as fne:
        print(f'Error: {fne}')

read_file('data.txt')
```

```python
##Assignment3: Write a function that takes a list of integers and returns their sum. Use try, except, and finally blocks to handle TypeError if a non-integer value is encountered and print an appropriate message.
def return_sum(lst):
    try:
        s = sum(lst)
        print(s)
    except TypeError as te:
        print(te)
return_sum([1,2,3,4,'a'])
```

```python
##Assignment5: Write a function that takes a dictionary and a key as input and returns the value associated with the key. Use try, except, and finally blocks to handle KeyError if the key is not found in the dictionary and print an appropriate message.
def key_error(dct,key):
    try:
        print(dct[key])
    except KeyError as ke:
        print(ke)
    finally:
        print("This is the output")
dct = {1:'a',2:'b'}
key_error(dct,1)
```

```python
##Assignment9: Write a function that attempts to parse a JSON string. Use try, except, and finally blocks to handle JSONDecodeError if the string is not a valid JSON and print an appropriate message.
import json
def parsing_json(json_string):
    try:
        data = json.loads(json_string)
        print(data)
    except json.JSONDecodeError as e:
        print(f'the error in the file is: {e}')
    finally:
        print("Execution Completed")

parsing_json('{"name": "kartikey", "age": 21}')
```

```python
##Assignment 10: Define a custom exception named `NegativeNumberError`. Write a function that raises this exception if a negative number is encountered in a list. Use try, except, and finally blocks to handle the custom exception and print an appropriate message.
class NegativeNumberError(Exception):
    pass
def custom_exception(l):
    try:
        for i in l:
            if i < 0:
                raise NegativeNumberError(f'The number {i} is negative.')
        print("all numbers are positive")
    except NegativeNumberError as nne:
        print(nne)
    finally:
        print("Execution Completed")

custom_exception([1,2,3,4,5,-6])
```

```python
##Assignment11: Write a function that calls another function which may raise an exception. Use try, except, and finally blocks to handle the exception and print an appropriate message.
class Errorr(Exception):
    pass
def call_function():
    raise Errorr('call_funciton was called')
def function_call():
    try:
        call_function()
    except Exception as e:
        print(f'Error: {e}')
    finally:
        print("Execution Completed")
function_call()
```

```python
##Assignment12: Define a class with a method that performs a division operation. Use try, except, and finally blocks within the method to handle division by zero and print an appropriate message.
class Calci():
    def division(a,b):
        try:
            result = a/b
        except ZeroDivisionError as e:
            print(f'Error: {e}')
            result = None
        finally:
            print("Execution Completed")
        return result

Calci.division(9,0)
    
```

```python
##Assingment 13: Write a function that takes a list of strings and converts them to integers. Use try, except, and finally blocks to handle ValueError if a string cannot be converted and print an appropriate message.
def convert_to_integer(s):
    try:
        i = int(s)
    except ValueError as ve:
        print(f'Error: {ve}')
        i = None
    finally:
        print(f'The integer is: {i}')
        print("Execution Completed")
convert_to_integer("123@")
```

```python
##Assignment14: Write a function that uses a list comprehension to convert a list of strings to integers. Use try, except, and finally blocks within the list comprehension to handle ValueError and print an appropriate message.
def convert_to_integer_2(lst):
    try:
        i = [int(k) for k in lst]
    except ValueError as ve:
        i = None
        print(f'Error: {ve}')
    finally:
        print(i)
convert_to_integer_2(["1","23","45"])
```

```python
##Assignment15: Write a function that attempts to write a list of strings to a file. Use try, except, and finally blocks to handle IOError and ensure the file is properly closed.
def write_in_file(lst):
    try:
        for i in lst:
            with open('file1.txt','a') as f:
                f.write(i + "\n")

    except IOError as io:
        print(f'Error: {io}')
    finally:
        print("Execution Completed")
write_in_file(["hello","world","this","is","not","real"])
```
