```python
##Assignment6: Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.
def replace_word():
    with open('data.txt', 'r') as file:
        content = file.read()

    old_word = input("Enter the word to replace: ")
    new_word = input("Enter the new word: ")

    content = content.replace(old_word, new_word)

    with open('data.txt', 'w') as file:
        file.write(content)

    print("Word replaced successfully.")

replace_word()
```

```python
##Assignment7: Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.
def reverse_print(filename):
    content = []
    with open(filename,'r') as file:
        content = file.read()
        content = content.split()
        content = content[::-1]
        for i in content:
            print(i)

reverse_print('reverse.txt')
```

```python
##Assignment8: Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.
def function1(filename):
    lines = 0
    words = 0
    char = 0
    with open(filename, 'r') as file:
        for l in file:
            char += len(l)
            lines += 1
            l = l.split()
            words += len(l)
        print(f'The number of lines are: {lines}, The number of words are: {words}, The number of chars are: {char}')
    
function1('stats.txt')
```

```python
##Assignment9: Write a function that merges the contents of multiple files into a single file named `merged.txt`.
def merge_files(*args):
    with open('merged.txt','w') as output:
        for file in args:
            with open(file,'r') as inputfile:
                output.write(inputfile.read())
        print("Files merged")
merge_files('data.txt','log.txt','stats.txt')
```
