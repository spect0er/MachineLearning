```python
## Assignment 9: Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.
def transpose_matrix(matrix):
    transposed = []
    for j in range(3):
        lst_1 = []
        for i in range(3):
            lst_1.append(matrix[i][j])
        transposed.append(lst_1)
    return transposed
matrix=[[1,2,3],
        [4,5,6],
        [7,8,9]]
transposed = transpose_matrix(matrix)
print(transposed)
```

```python
## Assignment 10: Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.
def flatten_list(nested_list):
    flatten = [nested_list[i][j] for i in range(len(nested_list)) for j in range(len(nested_list[0]))]
    return flatten
matrix=[[1,2,3],
        [4,5,6],
        [7,8,9]]
flat_list = flatten_list(matrix)
print(flat_list)
```
