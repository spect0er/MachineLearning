**Question 5:** Write a Python program to calculate the factorial of a number.

```python
# Your code here
def factorial(n):
    if n <= 1:
        return 1
    else:
        return n * factorial(n-1)

a = int(input("Enter a number to calculate it's factorial: "))
print(f"factorial of {a} is : {factorial(a)}")
```
