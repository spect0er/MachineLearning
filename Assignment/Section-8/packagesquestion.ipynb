```python
##Assignment9: Use the `datetime` module to print the current date, calculate the date 100 days from today, and determine the day of the week for a given date.
todays_date = dt.datetime.today()
print(f"Today's date: {todays_date}")
future_date = todays_date + dt.timedelta(days=100)
print(f"Days after 100 days: {future_date}")
random_date = dt.datetime(2026,12,12)
print(f"day on the date: {random_date} is : {random_date.strftime('%A')}")
```

```python
##Assignment10: Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.
from random import randint, shuffle
lst = [randint(1,50) for i in range(5)]
print(lst)
shuffle(lst)
print(f"Shuffled list: {lst}")
```
